```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

import os

from src.extraction import get_category_members, get_wikitext
from src.preprocessing import preprocess_text
```

```python
boss_names = get_category_members("Боссы", "https://eldenring.fandom.com/ru/api.php")
boss_pages = [get_wikitext(name,"https://eldenring.fandom.com/ru/api.php") for name in boss_names if name != "Боссы"]
```

```python
os.makedirs('../data', exist_ok=True)

for page in boss_pages:
    with open(f'../data/{page["title"]}.md', "w", encoding="utf-8") as file:
        file.write(preprocess_text(page['text'], page['title']))
```
