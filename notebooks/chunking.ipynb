```python
import sys
from pathlib import Path

import pandas as pd

sys.path.append(str(Path.cwd().parent))

from src.splitter import splitter
```

```python
all_chunks = splitter.process_folder(Path('../data'))
```

```python

records = []
for doc in all_chunks:
    records.append({
        'text': doc.page_content,
        'length': len(doc.page_content),
        'name': doc.metadata.get('name'),
        'chapter': doc.metadata.get('chapter'),
        'section': doc.metadata.get('section'),
        'subsection': doc.metadata.get('subsection')
    })

chunks_df = pd.DataFrame(records)
chunks_df = chunks_df[['text', 'length', 'name', 'chapter', 'section', 'subsection']]

chunks_df.head()
```

```python
def get_header_length(row):
    parts = [str(row['name']), str(row['chapter']), str(row['section']), str(row['subsection'])]
    parts = [p for p in parts if p != 'nan']
    return len("; ".join(parts) + "\n")

chunks_df['headers_length'] = chunks_df.apply(get_header_length, axis=1)
```
